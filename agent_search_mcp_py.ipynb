```python
from web_search_orchestrator import MCPAgentOrchestrator

orchestrator = MCPAgentOrchestrator(
    model_name="gemma4:12b-mlx", temperature=0.7
)

# Bootstrap tools and runtime graphs
await orchestrator.initialize()

test_queries = [
    "Winner of the 2026 Scripps National Spelling Bee?",
    "Distance from the earth to the moon",
    "When is Taylor Swift's birthday?",
    "In which city was the first match of the 2026 Fifa world cup played?",
    "Describe the carbon cycle",
]

print("--- Initiating Class-Based Agent Evaluation Loop ---\n")

for query in test_queries:
    result = await orchestrator.answer_query(query)

    # Pull the ultimate terminal text message payload
    final_response = result["messages"][-1].content
    tools_used = orchestrator.get_tool_list(result)

    print(f"Query: {query}")
    print(f"Answer: {final_response}")
    print(f"Tools Called: {tools_used}")
    print("-" * 40 + "\n")

```
